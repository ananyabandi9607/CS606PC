```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
url = "https://books.toscrape.com/"
response = requests.get(url)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text,"html.parser")
print(type(soup()))
```

```python
section = soup.find("section")
section
```

```python
articles = section.find_all("article")
articles
```

```python
data = []

for book in articles:

    book_info = {}

    name = book.find("h3")
    price = book.find("p", class_="price_color")
    avail = book.find("p", class_="instock availability")
    rating = book.find("p", class_="star-rating")

    book_info["Book Name"] = name.text.strip()
    book_info["Book Price"] = price.text.strip()[1:]
    book_info["Stock Availability"] = avail.text.strip()
    book_info["Rating"] = rating["class"][1]

    data.append(book_info)

df = pd.DataFrame(data)

student_info = ["Ananya Bandi", "24AG1A05D2", "CSE", "C", "DA"]


for i in range(len(student_info), len(data)):
    student_info.append("NA")

data_new = []

for j in range(len(data)):
    d = data[j]
    d["Student Details"] = student_info[j]
    data_new.append(d)

df = pd.DataFrame(data_new)

df = df[["Student Details",
           "Book Name",
           "Book Price",
           "Stock Availability",
           "Rating"]]

df
```

```python
df.to_csv("books.csv", index=False)
```
